# Data Cleaning
* remove estimated_year_of_production
* drop missing rows in repair_cafe_number, repair_cafe_name, kind_of_product, category, brand, repaird, repairability

```python
df.isna().sum()
```

```python
df['repaired'].value_counts()
```

```python
df['repaired'] = df['repaired'].replace("ja", "yes")
```

```python
df = df.drop(["repair_id", "estimated_year_of_production"], axis=1)
```

```python
df = df.dropna(subset=["repair_cafe_number", "repair_cafe_name", "category", "repaired", "kind_of_product", "brand", "repairability"])
```

```python
df.shape[0]
```

# Handling Data Types

```python
df['repair_cafe_number'] = df['repair_cafe_number'].apply(lambda x: str(int(x)))
```

```python
df = df.reset_index().drop("index", axis=1)
```

# Saving Data

```python
df.to_csv("../data/clean/data.csv", index_label="index")
```
