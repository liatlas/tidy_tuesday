```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
```

```python
datapath = "../data/clean/data.csv"
df = pd.read_csv(datapath, index_col="index")
```

```python
df.shape[0]
```

```python
df['repaired'].value_counts(normalize=True)
```

```python
sns.histplot(df['repairability'], bins=10)
plt.title("Distribution of Repairability")
plt.xlabel("Repairability")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.barplot(df['repair_cafe_name'].value_counts().head(25), orient="h")
```

```python
sns.barplot(df["kind_of_product"].value_counts().head(25), orient="h")
```

```python
sns.barplot(df['category'].value_counts(), orient="h")
```
