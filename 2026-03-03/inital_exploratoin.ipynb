```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
clutch_size = pd.read_csv("./data/clutch_size_cleaned.csv")
body_condition = pd.read_csv("./data/tortoise_body_condition_cleaned.csv")
```

```python
body_condition.head()
```

```python
fig, ax = plt.subplots()

sns.countplot(x='season', data=body_condition, ax=ax)

plt.show()
```

```python
fig, ax = plt.subplots()

sns.countplot(x='sex', data=body_condition, ax=ax)

plt.show()
```

```python
body_condition['locality'].value_counts()
```

```python
fig, ax = plt.subplots(figsize = (12,12))

sns.scatterplot(x="body_mass_grams", y="straight_carapace_length_mm", data=body_condition, hue="locality", alpha = 0.5,ax=ax)
ax.set_title("Body Mass vs. Carapace Length by Locality")
ax.set_xlabel("Body Mass (g)")
ax.set_ylabel("Straight Carapace Length (mm)")
```
